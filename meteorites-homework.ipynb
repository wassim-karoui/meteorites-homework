- How many entries are in the dataset?

Task is possible with jq/yq (`yq '. | length' meteorite_landings.json`) or Python dictionary processing but I'm using Pandas for convenience

````python
import pandas as pd
df = pd.read_json("meteorite_landings.json")
df.shape # returns number of rows and columns
````

# Answer is 1000

````python
# Viewing data as entry
df.sample(5)
````

````python
# Checking if all fields were assigned the right type, if not, I would do it manually
df.dtypes
````

````python
# It seems fine for mass, I would convert years to datetime64 then to integer
df["year"] = pd.to_datetime(df["year"]).dt.year.astype("int64")
````

````python
# There's a null value, let's fill it with zero
df["year"] = df["year"].fillna("01-01-01T00:00:00.000")
````

````python
# Starting over
df["year"] = pd.to_datetime(df["year"], format="mixed").dt.year.astype("int64")
df["year"]
````
